# Stage 1 — Environment + Data Prep

This notebook mounts Google Drive and runs the reproducible SciFact preparation script. It performs no baseline, training, or evaluation work. The repository must be cloned into `/content/RAFT` before running; change `REPO_DIR` if it is elsewhere.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from pathlib import Path

REPO_DIR = Path('/content/RAFT')
DRIVE_STAGE1_DIR = Path('/content/drive/MyDrive/RAFT/stage1_scifact')
SCRIPT = REPO_DIR / 'src' / 'prepare_data.py'
CONFIG = REPO_DIR / 'configs' / 'stage1_data_prep.json'
assert SCRIPT.is_file(), f'Clone the repository or update REPO_DIR: {SCRIPT}'
print(f'Persistent outputs: {DRIVE_STAGE1_DIR}')
```

```python
import subprocess

command = [
    'python', str(SCRIPT),
    '--config', str(CONFIG),
    '--raw-dir', str(DRIVE_STAGE1_DIR / 'raw'),
    '--output-dir', str(DRIVE_STAGE1_DIR / 'processed'),
]
subprocess.run(command, check=True)
```

All reusable artifacts and the source archive are now stored in Google Drive. Re-running the notebook uses the cached archive. Later stages should read the `processed` files directly.
